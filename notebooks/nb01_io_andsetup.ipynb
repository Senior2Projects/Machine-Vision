Cell 1 — imports

```python
import numpy as np
import matplotlib.pyplot as plt
import tkinter as tk
from tkinter import filedialog

from minicv.io import read_image, save_image
from minicv.color import rgb_to_gray, gray_to_rgb
```

Cell 2 — image picker

```python
def pick_image():
    root = tk.Tk()
    root.withdraw()
    root.attributes("-topmost", True)
    path = filedialog.askopenfilename(
        title="Select an image",
        filetypes=[("Image files", "*.png *.jpg *.jpeg")]
    )
    root.destroy()
    return path

IMAGE_PATH = pick_image()
print(f"Selected: {IMAGE_PATH}")
```

Cell 3 — read RGB and grayscale

```python
img_rgb  = read_image(IMAGE_PATH, mode="rgb")
img_gray = read_image(IMAGE_PATH, mode="gray")

print(f"RGB   shape: {img_rgb.shape},  dtype: {img_rgb.dtype}")
print(f"Gray  shape: {img_gray.shape}, dtype: {img_gray.dtype}")
```

Cell 4 - display both

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(img_rgb)
axes[0].set_title("RGB image")
axes[0].axis("off")

axes[1].imshow(img_gray, cmap="gray")
axes[1].set_title("Grayscale image")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

Cell 5 — color conversions

```python
gray_from_rgb  = rgb_to_gray(img_rgb)
rgb_from_gray  = gray_to_rgb(img_gray)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(gray_from_rgb, cmap="gray")
axes[0].set_title("rgb_to_gray()")
axes[0].axis("off")

axes[1].imshow(rgb_from_gray)
axes[1].set_title("gray_to_rgb()")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

Cell 6 — save and reload

```python
save_image(img_rgb,  "output_rgb.png")
save_image(img_gray, "output_gray.png")

reloaded_rgb  = read_image("output_rgb.png",  mode="rgb")
reloaded_gray = read_image("output_gray.png", mode="gray")

print(f"Saved and reloaded RGB  : {reloaded_rgb.shape}")
print(f"Saved and reloaded Gray : {reloaded_gray.shape}")
```
